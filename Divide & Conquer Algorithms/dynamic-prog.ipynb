Recursion    
fibonacci - f(n) = f(n-1) + f(n-2)   
Time complexity - O(2^n)     
Space complexity - O(1)

```python
def fibonacci(n):
    #Base cases 1,1,2
    if n <= 0:
        return None
    if n == 1:
        return 1
    if n == 2:
        return 1
    
    return fibonacci(n-1) + fibonacci(n-2) #repetitive subproblem
```

```python
fibonacci(8)
```

factorial = n! = n * (n-1)!

```python
def factorial(n):
    #Base cases
    if n == 0:
        return 1
    if n == 1:
        return 1
    return n * factorial(n-1)
```

```python
factorial(5)
```

Caching - Memoizing - saving solutions that we will use later    
Time complexity - O(n) - operation go through just the keys  
Space complexity - O(n)

Top-Down Dynammic Programming

```python
cache = dict()   #Key will be n, and the value will be the nth term
def fibo(n):
    #Base cases 1,1,2
    if n <= 0:
        return None
    if n == 1:
        return 1
    if n == 2:
        return 1
    
    #if it was already precomputed
    if n in cache.keys():
        return cache[n]
    
    cache[n] = fibo(n-1) + fibo(n-2) #solve and save first
    return cache[n]
```

```python
fibo(100)
```

Bottom-Up Dynamic Programming - only need to save 2 variables at a particular instance
Time complexity - O(n)    
Space complexity - O(1)

```python
def fib(n):
    #Base cases 1,1,2
    if n <= 0:
        return None
    if n == 1:
        return 1
    if n == 2:
        return 1
    
    #Iterative approach
    a, b = 1, 1
    for i in range(2, n):
        c = a + b
        a = b
        b = c
        
    return c
```

```python
fib(10000)
```
