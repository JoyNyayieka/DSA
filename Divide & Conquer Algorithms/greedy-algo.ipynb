Coin change problem

```python
def coin_change(coins, amount):
    coins.sort(reverse = True)     #[3,1,2] - [1,2,3] = [3,2,1]
    result = []
    for coin in coins:
        while amount >= coin:
            amount -= coin
            result.append(coin)
    return result
```

```python
coins =[1,5,10,25]
amount = 68
```

```python
coin_change(coins, amount)
```

```python
coins =[1,3,4]
amount = 13
coin_change(coins, amount)
```

Activity selection

```python
def selection(activities):  
    # A - (1-3)
    # B - (2-5)
    # C - (4-6)
    # D - (6-7)
    # E - (5-8)
    # F - (8-9)
    activities.sort(key = lambda x:x[1])  # takes activity as x, sorts it by the second character(end time)
    selected = []
    end = 0
    
    for start_time, end_time in activities:
        if start_time >= end:
            selected.append((start_time, end_time))
            end = end_time
    return selected
```

```python
activities = [(2,4), (1,5), (3,7), (8,10)]   #task B starts first
selection(activities)
```

```python
activities = [(3,4), (2,5), (6,7), (8,10)]   
selection(activities)
```
