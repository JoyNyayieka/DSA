Coin change problem

```python
def coin_change(coins, amount):
    coins.sort(reverse = True)     #[3,1,2] - [1,2,3] = [3,2,1]
    result = []
    for coin in coins:
        while amount >= coin:
            amount -= coin
            result.append(coin)
    return result
```

```python
coins =[1,5,10,25]
amount = 68
```

```python
coin_change(coins, amount)
```

```python
coins =[1,3,4]
amount = 13
coin_change(coins, amount)
```

Activity selection

```python
def selection(activities):  
    # A - (1-3)
    # B - (2-5)
    # C - (4-6)
    # D - (6-7)
    # E - (5-8)
    # F - (8-9)
    activities.sort(key = lambda x:x[1])  # takes activity as x, sorts it by the second character(end time)
    selected = []
    end = 0
    
    for start_time, end_time in activities:
        if start_time >= end:
            selected.append((start_time, end_time))
            end = end_time
    return selected
```

```python
activities = [(2,4), (1,5), (3,7), (8,10)]   #task B starts first
selection(activities)
```

```python
activities = [(3,4), (2,5), (6,7), (8,10)]   
selection(activities)
```

Fractional Knapsack problem - partial selections are allowed    
Highest value-to-weight ratio   
Example - thief in a jewelry store and you want to get the most valuable items for the max weight the bagg can carry

Input: val[] = [60, 100, 120]    
       wt[] =  [10, 20 , 30],   capacity = 50kg    
       v2w =   [6,  5,   4]    
Output: 240     
Explanation: We will take the items of weight 10kg and 20kg and 2/3 fraction of 30kg.     
Hence total value will be 60 + 100 + (2/3) * 120 = 240.         


Input: val[] = [500],      
       wt[] = [30], capacity = 10           
       v2w = 16       
Output: 166.667        

```python
# fractionalKnapsack
#
# The function is expected to return a FLOAT.
# The function accepts following parameters:
#  1. INTEGER W
#  2. INTEGER_ARRAY w
#  3. INTEGER_ARRAY p
# | Types | A | B | C | D |
# |-------|---|---|---|---| 
# Profits | 2 | 4 | 7 | 10 | 
# Weights | 1 | 3 | 5 | 7 |   If weight W =12,retun 17.6
#  p2w    | 2 | 1.33 | 1.40 | 1.42
#
# 12 - 1 = 11 - 7 = 4 
#      2  +    10  + (4/5*7) = 2+10+5.6 = 17.6

def fractionalKnapsack(W, w, p):
    n = len(p)
    
    items = [[p[i], w[i]] for i in range(n)]
    
    items.sort(key=lambda x:x[0]/x[1], reverse=True)
    
    res = 0.0
    currentCapacity = W
    
    for i in range(n):
        if items[i][1] <= currentCapacity:
            res += items[i][0]
            currentCapacity -= items[i][1]
        else:
            res += (1.0 * items[i][0] / items[i][1]) * currentCapacity
            
            break 
            
    return res

fractionalKnapsack(12, [1,3,5,7], [2,4,7,10])
```
