Bubble sort - time complexity of O(n^2), when array is small

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n - 1):
        for j in range(n-1-i):     #n-1-i because the last element is already in place
            if arr[j] > arr[j + 1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

arr = [5,1,6,3,7,8]
bubble_sort(arr)
```

Selection sort - arranges from highest to lowest or vv, select the first element, then follow with the rest - time complexity is O(n^2)

```python
def selection_sort(arr1):
    n = len(arr1)
    
    for i in range(n):
        start = i
        for j in range(i+1, n):      
            if arr1[j] < arr1[start]:
                start = j             #because we are comparing adjacent elements, not from the first index
        arr1[i], arr1[start] = arr1[start], arr1[i]
    return arr1
    
arr1 = [9,8,4,6,1,2,3]
selection_sort(arr1)
```

Insertion sort - assume that the first element is already sorted - time complexity of 0(n^2)

```python
def insertion_sort(arr3):
    for i in range (1, len(arr3)):
        key = arr3[i]
        j = i - 1 
        while j >= 0 and arr3[j] > key:
            arr3[j + 1] = arr3[j]
            j -= 1
        arr3[j + 1] = key 
    return arr3

arr3 = [11, 2, 3, 45, 55, 6, 36, 44]
print(insertion_sort(arr3))
```
