factorial of 5! -> 5*4*3*2*1    
factorial of 4! -> 4*3*2*1    
factorial of 5! -> 5*4!    

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n-1)
print(factorial(10))
```

fibonacci    
0,1,1,2,3,5,8,13,21     
0,1,2,3,4,5,6,7,8 - indexes  

fibonacci 13 = fibonacci 8 + fibonacci 5      
fibonacci n = fibonacci(n-1) + fibonacci(n-2)   

```python
def fibo(n):       #correct fibonacci function
    if n < 1:
        return None
    if n == 1:
        return 1
    if n == 2:
        return 1
    return fibo(n-1) + fibo(n-2)

print(fibo(7))
```

Merge sort - halfing the elements

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]
    
    left_sorted = merge_sort(left)
    right_sorted = merge_sort(right)
    
    return merge(left_sorted, right_sorted)

def merge(left, right):
    result = []
    i = j = 0 # pointer for left and right respectively
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

print(merge_sort([2,13,5,7,88,45,62,56]))
```

Quick sort - use pivots to divide element

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return (arr)
    
    pivot = arr[-1]
    left = []
    right = []
    equal = []
    
    for num in arr:
        if num < pivot:
            left.append(num)
        elif num == pivot:
            equal.append(num)
        else:
            right.append(num)
            
    return quick_sort(left) + equal + quick_sort(right)

print(quick_sort([38,27,44,56,13,28,5,99]))
```
