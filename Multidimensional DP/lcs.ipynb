Longest Common Subsequence(LCS)    
Subsequence - a sequence that can be derived from another sequence by deleting some elements without changing the order of the remaining elements     
Substring

```python
#top down approach
cache = dict()
def lcs(a,b):
    #Base cases
    if len(a) == 0 or len(b) == 0:
        return 0    
    if len(a) == 1 and len(b) == 1:
        return 1 if a == b else 0
    
    if (a,b) in cache.keys():
        return cache[(a,b)]
        
    #Repetitive subproblem
    if a[0] == b[0]:
        cache[(a,b)] = 1 + lcs(a[1:], b[1:])
        return cache[(a,b)]
    
    cache[(a,b)] =  max(lcs(a[1:], b), lcs(a, b[1:]))
    return cache[(a,b)]
```

```python
lcs("abcdefghi","ecdgi")
```

```python
#bottom up approach - grid logic - if the letters are similar, diagonal + 1,
#if letters are not similar, top and left, select larger one
def lcs2(a,b):
    #Base cases
    if len(a) == 0 or len(b) == 0:
        return 0    
    if len(a) == 1 and len(b) == 1:
        return 1 if a == b else 0
    
    prev = [0] * (len(a) + 1)

    for char in b:
        curr = [0] * (len(a) + 1)
        for i in range(1, len(a)+1):
            if char == a[i-1]:
                curr[i] = 1 + prev[i-1]
            else:
                curr[i] = max(curr[i-1], prev[i])
                
        prev = curr
        
    return curr[-1]
```

```python
lcs2("abcdefghi","ecdgi")
```

          Run time     Space complexity     
Recursion O(2^n)            O(1)      
Top down  O(n.m)            O(n.m)      
Bottom up   O(n.m)            O(n)
