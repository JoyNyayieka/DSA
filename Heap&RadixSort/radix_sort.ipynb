Time complexity  -> O(nk) n is the number of elements and k is the number of digits.
Space complexity -> O(n+b) b is the base

```python
def counting_sort(arr, place):
    n = len(arr)
    output = [0] * n    #[0,0,0,0,n times]
    count = [0] * 10    #buckets from 0 to 9
    for i in range(n):
        digit = (arr[i] // place) % 10 
        count[digit] += 1
    for i in range(1, 10):
        count[i] += count[i - 1]
        
    i = n - 1
    while i >= 0:
        digit = (arr[i] // place) % 10 
        output[count[digit] - 1] = arr[i]
        count[digit] -= 1
        i -= 1 
    for i in range(n):
        arr[i] = output[i]
        
def radix_sort(arr):
    max_num = max(arr)
    place = 1
    while max_num // place > 0:
        counting_sort(arr, place)
        place *= 10 
```

```python
arr = [170, 45, 75, 90, 802, 24, 2, 66]
radix_sort(arr)
arr
```

```python
def rob(self, nums):
        """
        :type nums: List[int]
        :rtype: int
        """
        cache = dict()
        #Base case
        if len(nums) == 1:
            return nums[0]
        if len(nums) == 2:
            return max(nums[0], nums[1])

        if tuple(nums) in cache.keys():
            return cache[tuple(nums)]

        #Repetitive Subproblem
        a = nums[-1] + rob(nums[:-2])
        b = rob(nums[:-1])
        cache[tuple(nums)] = max(a,b)
        return max(a, b)
```
