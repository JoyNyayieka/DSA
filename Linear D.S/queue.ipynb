Queues - FIFO - First in, first out

```python
class Queue:
    def __init__(self):
        self.queue = []
        
    def enqueue(self, data):
        self.queue.append(data)
        print(f"{data} has been added at the end of the line.")
        
    def dequeue(self):
        if self.isEmpty():
            print("The queue is empty.")
            return
        removed = self.queue.pop(0)
        print(f"{removed} has been removed from the queue.")
        
    def peek(self):
        if self.isEmpty():
            print("Queue is empty.")
            return 
        print(f"Current element is:  {self.queue[0]}")
    def isEmpty(self):
        if len(self.queue) == 0:
            print("Queue is empty.")
    def display(self):
        print("Queue from the front to rear:")
        for item in self.queue:
            print(item)
```

```python
Q = Queue()
Q.enqueue(1)
Q.enqueue(2)
Q.enqueue(3)
Q.enqueue(4)
Q.enqueue(5)
```

```python
Q.dequeue()
```

```python
Q.display()
```
