Unbounded knapsack - an item can be chosen a couple of times   
Qs - minimum number of coins, max, how many different ways can you give change

Top-down DP for Knapsack Problem

```python
def coinchange(amount, coins):
    #Base cases
    if amount == 0:
        return 1 #valid
    if len(coins) == 0 and amount > 0:
        return 0 #Not a valid way
    if len(coins) == 1:
        if coins[0] > amount:
            return 0 
        if amount%coins[0] != 0:
            return 0
        return 1
         
    #Repatitive sub-problem
    if coins[-1] <= amount:
        a = coinchange(amount-coins[-1], coins)
        b = coinchange(amount, coins[:-1])
        return a + b
    return coinchange(amount, coins[:-1])
```

```python
coinchange(7, [1,3,4,5])
```

```python
#Adding memoisation

cache = dict()

def coinchange(amount, coins):
    #Base cases
    if amount == 0:
        return 1 #valid
    if len(coins) == 0 and amount > 0:
        return 0 #Not a valid way
    if len(coins) == 1:
        if coins[0] > amount:
            return 0 
        if amount%coins[0] != 0:
            return 0
        return 1
        
    if (amount, tuple(coins)) in cache.keys():
        return cache[(amount, tuple(coins))]
    
    #Repetitive sub-problem
    if coins[-1] <= amount:
        a = coinchange(amount-coins[-1], coins)
        b = coinchange(amount, coins[:-1])
        cache[(amount, tuple(coins))] = a + b
        return cache[(amount, tuple(coins))]
    return coinchange(amount, coins[:-1])
```

```python
coinchange(7, [1,3,4,5])
```

Bottom up approach

```python
#minimum coins
def change(amount, coins):
    #Base case
    if amount == 0:
        return 0
    if len(coins) == 0 and amount > 0:
        return float('inf')
    
    #Repetitive subproblem
    prev = [0] + amount * [float('inf')]
    
    for coin in coins:
        curr = [0] * (amount + 1)
        for i in range(amount + 1 ):
            if coin > i:
                curr[i] = prev[i]
            else:
                a = prev[i]
                b = curr[i-coin]+1
                curr[i] = min(a, b)
        #print(prev)
        prev = curr
    
    return curr[-1]
```

```python
change(11, [1,3,5,7])
```

```python
#how many different ways
def getWays(n, c):
    # ways[x] = number of ways to make amount x
    ways = [0] * (n + 1)
    ways[0] = 1  # base case

    # Build up from smaller amounts to n
    for coin in c:
        for amt in range(coin, n + 1):
            ways[amt] += ways[amt - coin]

    return ways[n]
```

```python
getWays(4, [2,1])
```
