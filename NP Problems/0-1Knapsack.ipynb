0-1 Knapsack problem - items cannot be divided eg people, you cannot half a person to fit into an elevator    
Items can only be chosen once

Top-down Dynamic Programming approach

```python
def knapsack(maxW, w, p):
    #Base cases
    if len(w) == 0:
        return 0
    if len(w) == 1:
        if w[0] <= maxW:
            return p[0]
        return 0
    
    if w[-1] <= maxW:
        include = p[-1] + knapsack(maxW - w[-1], w[:-1], p[:-1])
        exclude = knapsack(maxW, w[:-1], p[:-1])
        
        return max(include, exclude)
    
    else:
        return knapsack(maxW, w[:-1], p[:-1])
```

```python
maxW = 8
w = [1,3,5,7]
p = [2,4,7,10]
knapsack(maxW, w, p)
```

Memoization

```python
cache = dict()

def knapsack(maxW, w, p):
    #Base cases
    if len(w) == 0:
        return 0
    if len(w) == 1:
        if w[0] <= maxW:
            return p[0]
        return 0
    
    if (tuple(w), maxW) in cache.keys():
        return cache[(tuple(w), maxW)]
    
    if w[-1] <= maxW:
        include = p[-1] + knapsack(maxW - w[-1], w[:-1], p[:-1])
        exclude = knapsack(maxW, w[:-1], p[:-1])
        cache[tuple(w), maxW] = max(include, exclude)
        return cache[tuple(w), maxW]
    else:
        cache[tuple(w), maxW] = knapsack(maxW, w[:-1], p[:-1])
        return cache[tuple(w), maxW]
```

```python
maxW = 8
w = [1,3,5,7]
p = [2,4,7,10]
knapsack(maxW, w, p)
```
