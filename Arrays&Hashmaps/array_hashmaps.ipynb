Linear Traversal

```python
arr = [2,3,45,23,1,3,5,7,8,65,34]
print("linear traverse", end=" ")
for i in arr:
    print(i, end=" ")
print()
```

```python
#Reverse traversal
arr = [2,3,45,23,1,3,5,7,8,65,34]
print("reverse traverse", end=" ")
for i in range(len(arr)-1, -1, -1):
    print(arr[i], end=" ")
print()
```

```python
arr1 = [2,3,45,5,23,1,3,5,7,8,65,5,34]
print("Reverse traversal", end=" ")
for i in range(len(arr1)-1, -1, -1):
    print(arr1[i], end=" ")
print()
```

```python
value_to_delete = 5 
arr1 = [x for x in arr1 if x != value_to_delete] 
print(arr1) 
```

Hashmaps -> dictionary, key value pairs, stored in an unordered way therefore cannot be indexed

```python
scores = {1:85, 2:90, 3:78, 4:92}
scores[2]
```

```python
print("\nKey–Value pairs:") 
for key, value in scores.items(): 
    print(key, ":", value) 
```

```python
scores[4] = 88  #inserting elements
```

```python
scores
```

```python
scores[2] = 20 #updating elements
scores
```

```python
scores = {1: 85, 2: 90, 3: 78} 
if 2 in scores: 
    print("Key found! Value =", scores[2]) 
else: 
    print("Key not found") 
```

```python
del scores[2]
scores
```

```python
scores.clear()
scores
```
