Stacks - LIFO - Last In First Out

```python
class Stacks:
    def __init__(self):
        self.stack = []
        
    def push(self, data):
        self.stack.append(data)
        print(f"{data} added to the stack")
    
    def pop(self):
        if self.isEmpty():
            print("stack is empty")
            return
        removed = self.stack.pop()
        print(f"{removed} has been removed")
        
    def peek(self):
        if self.isEmpty():
            print("stack is empty")
            return
        print(f"top element is {self.stack[-1]}")
        
    def isEmpty(self):
        return len(self.stack) == 0
    
    def display(self):
        print("stack top to bottom ")
        for item in reversed(self.stack):
            print(item)
            
stack = Stacks()
stack.push(1)     
stack.push(2)  
stack.push(3)  
stack.push(4)  
stack.push(5)  
    
```

```python
stack.peek()
```

```python
stack.pop()
```
